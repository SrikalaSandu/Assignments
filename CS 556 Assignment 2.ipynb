## 1. Solving System of Linear Equations

### 1.1 Row Reduction approach

```python
import numpy as np

# Helper functions
def AddRows(M, src_row, dest_row, scalar):
    """
    Perform row operation: dest_row = dest_row + scalar * src_row
    """
    Mnew = M.copy()
    Mnew[dest_row] += scalar * Mnew[src_row]
    return Mnew

def MultiplyRow(M, row, scalar):
    """
    Multiply a row by a scalar: row = scalar * row
    """
    Mnew = M.copy()
    Mnew[row] *= scalar
    return Mnew
```

Solve the following system of linear equations using the reduction method (as mentioned in the week 2 lab):
$$\begin{cases}
x + 2y - 3z + 4w = 12, \\ 2x + 2y - 2z + 3w = 10, \\ y + z = -1, \\ x - y + z - 2w = -4 \end{cases}\tag{1}$$


```python
print("Problem 1.1: Row Reduction Approach")

# Define matrix A and vector b for system Ax = b
A = np.array([
    [1, 2, -3, 4],
    [2, 2, -2, 3],
    [0, 1, 1, 0],
    [1, -1, 1, -2]
], dtype=float)

b = np.array([12, 10, -1, -4], dtype=float)

# Create augmented matrix
Aaug = np.hstack((A, b.reshape(-1,1)))

# Gaussian elimination forward phase
n = A.shape[0]

for i in range(n):
    # Make pivot = 1 by dividing row i with pivot element
    pivot = Aaug[i, i]
    Aaug[i] = Aaug[i] / pivot
    
    # Make zeros in rows below pivot
    for j in range(i+1, n):
        factor = -Aaug[j, i]
        Aaug = AddRows(Aaug, i, j, factor)

# Back substitution to get reduced row echelon form
for i in range(n-1, -1, -1):
    for j in range(i-1, -1, -1):
        factor = -Aaug[j, i]
        Aaug = AddRows(Aaug, i, j, factor)

print("Reduced Row Echelon Form:")
print(np.round(Aaug, 3))

# Solution vector is last column
solution = Aaug[:, -1]
print("Solution (Row Reduced):", solution)
```

### 1.2 `Numpy` linear algebra package
Solve the above system of linear equations using the `numpy.linalg` package.

```python
print("\nProblem 1.2: NumPy Linear Algebra Solve")

sol_np = np.linalg.solve(A, b)
print("Solution (NumPy):", sol_np)
```

## 2. Determinant
### 2.1 Compute the determinant of matrix $A$ using the `numpy.linalg` package:

$$ \begin{align}
  \textbf{A}  = \begin{bmatrix}
1 & 0 & -2\\
3 & 1 & -2\\
-5 & -1 & 9
\end{bmatrix}
  \end{align}
  $$

```python
B = np.array([
    [1, 2, 1],
    [4, 4, 5],
    [6, 7, 7]
], dtype=float)

print("\nProblem 2.1: Determinant of B")
det_B = np.linalg.det(B)
print("Determinant:", det_B)

print("\nProblem 2.2: Determinant Check for Invertibility")
if abs(det_B) > 1e-10:
    print("Matrix B is invertible.")
else:
    print("Matrix B is not invertible.")
```

### 2.2 Is the above matrix invertible? Why?

A matrix can be invertible if its determinant is a nonzero. From the above solution we can say matrix is invertible because the determinant is non zero which is the determinant of matrix is 3 which is not equal to zero.

# 3. Matrix Inverse
### 3.1 Compute the inverse of non-singular matrix $B$ using row reduction
$$ \begin{align}
  \textbf{B}  = \begin{bmatrix}
1 & 2 & 1\\
4 & 4 & 5\\
6 & 7 & 7
\end{bmatrix}
  \end{align}
  $$

```python
print("\nProblem 3.1: Matrix Inverse by Row Reduction")

# Augment B with identity matrix I
n = B.shape[0]
Aaug = np.hstack((B, np.eye(n)))

# Perform row operations to convert left side to identity matrix
# Use AddRows and MultiplyRow functions

# Normalize row 0
Aaug = MultiplyRow(Aaug, 0, 1/Aaug[0,0])

# Make zeros below pivot in col 0
Aaug = AddRows(Aaug, 0, 1, -Aaug[1,0])
Aaug = AddRows(Aaug, 0, 2, -Aaug[2,0])

# Normalize row 1
Aaug = MultiplyRow(Aaug, 1, 1/Aaug[1,1])

# Make zeros above and below pivot in col 1
Aaug = AddRows(Aaug, 1, 0, -Aaug[0,1])
Aaug = AddRows(Aaug, 1, 2, -Aaug[2,1])

# Normalize row 2
Aaug = MultiplyRow(Aaug, 2, 1/Aaug[2,2])

# Make zeros above pivot in col 2
Aaug = AddRows(Aaug, 2, 0, -Aaug[0,2])
Aaug = AddRows(Aaug, 2, 1, -Aaug[1,2])

# Extract inverse matrix
Binverse = Aaug[:, n:]

print("Inverse matrix by row reduction:")
print(np.round(Binverse, 3))
```

### 3.2 Compute the inverse matrix $B$ using the `numpy.linalg` package.

```python
print("\nProblem 3.2: Matrix Inverse Using NumPy")

Binv_np = np.linalg.inv(B)
print("Inverse matrix (NumPy):")
print(np.round(Binv_np, 3))
```
